#  Move All Zeros to the End

```python
def movezeros(num):
    i,j=0,0
    while i<len(num):
        if num[i]!=0:
            num[i],num[j]=num[j],num[i]
            i+=1
            j+=1
        else:
            i+=1
    return num
movezeros([0, 1, 0, 3, 12])

```

# Intersection of Two Arrays

```python
def intersection(num1,num2):
    num3=[]
    for i in num1:
        if i in num2:
            num3.append(i)
    return num3

intersection([1, 2, 3, 4, 5],[3, 4, 5, 6, 7])
```

# Maximum Difference

```python
def maxdiff(num):
    smallest=num[0]
    diff=0
    for i in range(1,len(num)):
        currentdif=num[i]-smallest
        if currentdif > diff:
            diff=currentdif
        if num[i] < smallest:
            smallest=num[i]
    return diff
```

# Remove Duplicates

```python
def removeduplicate(num):
    if len(num) == 0:
        return []

    i = 0

    for j in range(1, len(num)):
        if num[i] != num[j]:
            i += 1
            num[i] = num[j]

    return num[:i + 1]


print(removeduplicate([1, 1, 2, 2, 3, 4, 4]))
```
